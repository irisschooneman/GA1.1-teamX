```python
def multiply(a, b):
    return a * b
```

```python
multiply(5, 6)
```

```python
def square(x):
    return multiply(x, x)
```

```python
print(square(2))
print(square(5))
print(square(-3))
```

# use and missuse
fucntion only accepts int types, for example 5, 6. It does not accept strings, bool, etc.
